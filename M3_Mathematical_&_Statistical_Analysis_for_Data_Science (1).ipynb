##Task 1 – Linear Algebra Implementation

```python
import numpy as np

vector_a = np.array([2, 4, 6])
vector_b = np.array([1, 3, 5])
matrix_a = np.array([
    [2, 1, 3],
    [1, 0, 2],
    [4, 1, 1]
])
matrix_b = np.array([
    [1, 2, 0],
    [3, 1, 4],
    [2, 5, 1]
])

print("Vector Addition:", vector_a + vector_b)
print("Vector Subtraction:", vector_a - vector_b)
print("Dot Product:", np.dot(vector_a, vector_b))

print("\nMatrix Addition:\n", matrix_a + matrix_b)
print("\nMatrix Multiplication:\n", matrix_a @ matrix_b)

det_a = np.linalg.det(matrix_a)
print("\nDeterminant of matrix_a:", det_a)

if not np.isclose(det_a, 0):
    inv_a = np.linalg.inv(matrix_a)
    print("\nInverse of matrix_a:\n", inv_a)
else:
    print("Matrix is singular; inverse does not exist.")

eigenvalues, eigenvectors = np.linalg.eig(matrix_a)
print("\nEigenvalues:", eigenvalues)
print("\nEigenvectors:\n", eigenvectors)
```

```python
import numpy as np

sales_success = [0.65, 0.35]
customer_segments = ["Premium", "Regular"]

segment_purchase_probability = {
    "Premium": 0.80,
    "Regular": 0.50
}

probability_success = sum(sales_success[i] * segment_purchase_probability[customer_segments[i]]for i in range(len(customer_segments)))

print("Probability of successful sales:", probability_success)
print(f'')
print("Percentage:", probability_success * 100, "%")

p_purchase_given_premium = segment_purchase_probability["Premium"]
p_purchase_given_regular = segment_purchase_probability["Regular"]
print(f'')
print("p_purchase_given_premium", p_purchase_given_premium)
print(f'')
print("p_purchase_given_regular", p_purchase_given_regular)
```

#Task 3 – Descriptive Statistics

```python
import numpy as np
import pandas as pd
import statistics

sales_data = [
    120, 135, 150, 145, 160, 155, 170, 165, 180, 175,
    190, 200, 210, 195, 185, 175, 165, 155, 145, 135
]


mean_value = np.mean(sales_data)
median_value = np.median(sales_data)
mode_value = statistics.mode(sales_data)
sales_range = np.max(sales_data) - np.min(sales_data)
variance = np.var(sales_data)
standard_deviation = np.std(sales_data)
print(f"")
print("Mean", mean_value)
print(f"")
print("Median", median_value)
print(f"")
print("Mode", mode_value)
print(f"")

print("Range", sales_range)
print(f"")
print("Variance", variance)
print(f"")
print("Standard Deviation", standard_deviation)
print(f"")

sales = pd.Series(sales_data)
skewness = sales.skew()
kurtosis = sales.kurt()
print(f"")
print("Skewness", skewness)
print(f"")
print("Kurtosis", kurtosis)
print(f"")
```

**Interpret about dataset**

The sales dataset is approximately symmetric, because its skewness is very close to zero. The slight negative value indicates a very weak negative skew, but it is not substantial enough to consider the dataset strongly negatively skewed.
